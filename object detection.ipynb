```python
import torch
import cv2
import pyttsx3  # optional: to speak object names

# Load pre-trained YOLOv5 model
model = torch.hub.load('ultralytics/yolov5', 'yolov5s', pretrained=True)

# Initialize webcam
cap = cv2.VideoCapture(0)  # 0 for default webcam

# Initialize text-to-speech engine
engine = pyttsx3.init()

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Detect objects
    results = model(frame)

    # Parse results
    labels = results.xyxyn[0][:, -1].numpy()
    cords = results.xyxyn[0][:, :-1].numpy()
    names = results.names

    for i in range(len(labels)):
        x1, y1, x2, y2, conf = cords[i]
        label = int(labels[i])
        object_name = names[label]
        confidence = conf

        # Draw bounding box and label
        h, w, _ = frame.shape
        cv2.rectangle(frame, (int(x1 * w), int(y1 * h)), (int(x2 * w), int(y2 * h)), (255, 0, 0), 2)
        cv2.putText(frame, f'{object_name} {confidence:.2f}', (int(x1 * w), int(y1 * h) - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.9, (255, 255, 255), 2)

        # Speak the object name (optional)
        engine.say(object_name)
        engine.runAndWait()

    # Show result
    cv2.imshow('Object Detection', frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release everything
cap.release()
cv2.destroyAllWindows()
```
